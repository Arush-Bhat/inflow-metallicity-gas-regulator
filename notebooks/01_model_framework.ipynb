# 01 Model Framework

This notebook defines the baseline one-zone gas regulator used in Part I.

It contains the state variables, assumptions, evolution equations, adopted parameters, and analytical equilibrium relations. Numerical validation is done in Notebook 02.

## Assumptions and units

The baseline model uses:

- one well-mixed gas reservoir
- constant inflow rate during each run
- constant inflow metallicity during each run
- instantaneous stellar mass return
- outflow proportional to the star formation rate
- outflow metallicity equal to the gas metallicity
- constant gas depletion time

Mass is in $M_\odot$, time in Gyr, rates in $M_\odot\,{\rm Gyr}^{-1}$, and metallicity is a mass fraction.

## State variables

The model evolves

$$[M_g,\ M_\star,\ M_Z]$$

where:

- $M_g$ is the gas mass
- $M_\star$ is the mass locked in long-lived stars
- $M_Z$ is the metal mass in the gas

The gas metallicity is

$$
Z_g=\frac{M_Z}{M_g}.
$$

## Star formation and gas flows

The baseline star formation rate is

$$
\dot M_{\star,{\rm SF}}=\frac{M_g}{\tau_{\rm dep}}.
$$

The outflow rate is

$$
\dot M_{\rm out}=\eta\dot M_{\star,{\rm SF}},
$$

and the instantaneous returned rate is

$$
\dot M_{\rm return}=R\dot M_{\star,{\rm SF}}.
$$

Therefore the stellar mass retained in long-lived stars grows as

$$
\frac{dM_\star}{dt}=(1-R)\dot M_{\star,{\rm SF}}.
$$

Here $\dot M_{\star,{\rm SF}}$ denotes the total star formation rate, while $dM_\star/dt$ is the surviving stellar mass growth rate.

Reference: Lilly et al. (2013), *Gas Regulation of Galaxies: The Evolution of the Cosmic Specific Star Formation Rate, the Metallicity, and the Stellar Content of Halos*, ApJ, 772, 119.

## Evolution equations

Gas mass evolves as

$$
\frac{dM_g}{dt}=\dot M_{\rm in}-(1-R+\eta)\dot M_{\star,{\rm SF}}.
$$

Stellar mass evolves as

$$
\frac{dM_\star}{dt}=(1-R)\dot M_{\star,{\rm SF}}.
$$

Gas-phase metal mass evolves as

$$
\frac{dM_Z}{dt}
=Z_{\rm in}\dot M_{\rm in}
-Z_g(1-R+\eta)\dot M_{\star,{\rm SF}}
+y(1-R)\dot M_{\star,{\rm SF}}.
$$

Here $R$ is the return fraction, $\eta$ is the mass-loading factor, $Z_{\rm in}$ is the inflow metallicity, and $y$ is the metal yield per mass locked in long-lived stars.

These equations assume instantaneous recycling and $Z_{\rm out}=Z_g$.

References: Lilly et al. (2013), ApJ, 772, 119. Equivalent gas and metal conservation equations are given in Carton et al. (2015), MNRAS, 451, 210, equations 4 and 7.

## Adopted baseline parameters

The model uses literature-guided values together with representative initial conditions.

- $Z_\odot=0.0134$: adopted solar metal mass fraction from Asplund et al. (2009).
- $\tau_{\rm dep}=2.35$ Gyr: adopted one-zone depletion timescale. The value is motivated by the molecular-gas depletion timescale reported by Bigiel et al. (2011), while its application to the total one-zone gas reservoir is a modelling assumption. Gas-regulator fits in Lilly et al. (2013) also give total gas depletion timescales of order $2$--$3$ Gyr.
- $R=0.4$: instantaneous return fraction adopted by Lilly et al. (2013).
- $y=0.016$: metal yield per unit mass locked into long-lived stars, consistent with the value discussed by Lilly et al. (2013).
- $\eta=0.5$: representative mass-loading factor adopted for this model. It is not independently calibrated within this project.

$M_{g,0}$, $M_{\star,0}$ and $Z_{g,0}$ are representative initial conditions rather than measured values.

The inflow rate is defined by

$$
\dot M_{\rm in}=(1-R+\eta)\frac{M_{g,0}}{\tau_{\rm dep}},
$$

so the baseline model begins in gas-mass equilibrium.

Parameter references: Asplund et al. (2009), ARA&A, 47, 481; Bigiel et al. (2011), ApJL, 730, L13; Lilly et al. (2013), ApJ, 772, 119; Kudritzki et al. (2015), MNRAS, 450, 342.

## Analytical equilibrium

Gas equilibrium requires

$$
\frac{dM_g}{dt}=0.
$$

Therefore

$$
\dot M_{\star,{\rm SF,eq}}
=\frac{\dot M_{\rm in}}{1-R+\eta},
$$

and for the baseline star formation law

$$
M_{g,{\rm eq}}
=\tau_{\rm dep}\dot M_{\star,{\rm SF,eq}}.
$$

If the metal mass is also in equilibrium,

$$
\frac{dM_Z}{dt}=0,
$$

then using the gas-equilibrium relation gives

$$
Z_{\rm eq}
=Z_{\rm in}
+\frac{y(1-R)}{1-R+\eta}.
$$

These equilibrium expressions are derived directly from the evolution equations above.

```python
eq = equilibrium_values(BASE_PARAMS)

print(f"Equilibrium SFR = {eq['Mdot_star']/1e9:.3f} Msun/yr")
print(f"Equilibrium gas mass = {eq['M_g']:.3e} Msun")
print(f"Equilibrium metallicity = {eq['Z_g']:.6f}")
print(f"Equilibrium metallicity = {eq['Z_g']/Z_SUN:.3f} Z_sun")
```

## Parameter and assumption provenance

All physical parameters, initial conditions, controlled variables, derived
quantities, and numerical settings are documented separately in
`PARAMETER_PROVENANCE.md`.

This distinction is important because not every numerical value in the model
has the same evidential status. Some quantities are taken directly from the
literature, some are explicit modelling assumptions, some define the initial
state, and $Z_{\rm in}$ is the controlled experimental variable.
